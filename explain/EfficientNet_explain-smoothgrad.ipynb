```python
from utils import visualise_smoothgrad
import tensorflow as tf
import re
```

```python
# Load and preprocess the images
# Mild
mild_path = '../archive/gaussian_filtered_images/Mild/0024cdab0c1e.png'

# Moderate
moderate_path = '../archive/gaussian_filtered_images/Moderate/000c1434d8d7.png'

# No_DR
# no_dr_path = '../archive/gaussian_filtered_images/No_DR/002c21358ce6.png'
no_dr_path = '../archive/gaussian_filtered_images/No_DR/05b1bb2bdb81.png'

# Proliferate_DR
# prof_dr_path = '../archive/gaussian_filtered_images/Proliferate_DR/001639a390f0.png'
prof_dr_path = '../archive/gaussian_filtered_images/Proliferate_DR/02dda30d3acf.png'

# Severe
severe_path = '../archive/gaussian_filtered_images/Severe/0104b032c141.png'
```

```python
# Load the VGG16 model
model_path = '../models/efficientnet_model.keras'
model = tf.keras.models.load_model(model_path)
```

```python
# Isolate the layers that Grad-CAM can utilise
conv_layer_name_list = []
for layer in model.layers:
    if bool(re.search(r'conv$', layer.name)):
        conv_layer_name_list.append(layer.name)
    
# print(conv_layer_name_list)
print(len(conv_layer_name_list))
```

```python
# Visualising No DR Case
print("NO DR")
visualise_smoothgrad("efficientnet", model, no_dr_path, conv_layer_name_list, rows=5, cols=5, mod_factor=2)
```

```python
# Visualising Mild Case
print("MILD")
visualise_smoothgrad("efficientnet", model, mild_path, conv_layer_name_list, rows=5, cols=5, mod_factor=2)
```

```python
# Visualising Moderate Case
print("MODERATE")
visualise_smoothgrad("efficientnet", model, moderate_path, conv_layer_name_list, rows=5, cols=5, mod_factor=2)
```

```python
# Visualising Proliferate DR Case
print("PROLIFERATE DR")
visualise_smoothgrad("efficientnet", model, prof_dr_path, conv_layer_name_list, rows=5, cols=5, mod_factor=2)
```

```python
# Visualising Severe Case
print("SEVERE")
visualise_smoothgrad("efficientnet", model, severe_path, conv_layer_name_list, rows=5, cols=5, mod_factor=2)
```
